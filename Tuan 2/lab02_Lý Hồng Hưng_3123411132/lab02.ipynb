## Bài 1. Tạo Tensor cơ bản
Tạo Tensor số nguyên kích thước `2 × 3` và kiểm tra kích thước, số chiều, kiểu dữ liệu.

```python
import torch
a = torch.tensor(
[[1, 2, 3],
[4, 5, 6]],
dtype=torch.int32
)
print("Tensor a:")
print(a)
print("Kích thước:", a.shape)
print ("Chiều:", a.ndim)
print("Kiểu dữ liệu:", a.dtype)
```

## Bài 4. Slicing và Indexing

Truy xuất dữ liệu từ Tensor nhiều chiều và lấy hai phần tử cuối cùng trên chiều thứ ba.

## Bài 5. Reshape và Flatten

Thay đổi hình dạng Tensor nhưng không thay đổi số phần tử.

```python
import torch

a = torch.randn(3, 4)

print("Shape ban đầu:", a.shape)
print("Số phần tử ban đầu:", a.numel())

b = a.ravel()
print("\nFlatten:")
print(b)
print("Shape:", b.shape)

c = a.reshape(3, 2, 2)
print("\nReshape 3 x 2 x 2:")
print(c)

d = a.reshape(2, 6)
print("\nReshape 2 x 6:")
print(d)
print("Shape:", d.shape)
print("Số phần tử sau reshape:", d.numel())
```

## Bài 6. Phép toán và nhân ma trận

Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a * b:")
print(a * b)

print("\na @ b.T:")
print(a @ b.T)

print("Shape của a * b:", (a * b).shape)
print("Shape của a @ b.T:", (a @ b.T).shape)
```

## Bài 7. Các hàm thống kê trên Tensor

Tính trung bình, độ lệch chuẩn và tổng tích lũy theo các chiều khác nhau.

```python
import torch

a = torch.randn(3, 4)

print("Tensor a:")
print(a)

print("\nMean theo cột:")
print(torch.mean(a, dim=0))

print("\nĐộ lệch chuẩn theo cột:")
print(torch.std(a, dim=0))

print("\nTổng tích lũy theo cột:")
print(torch.cumsum(a, dim=0))

print("\nMean theo hàng:")
print(torch.mean(a, dim=1))

print("\nĐộ lệch chuẩn theo hàng:")
print(torch.std(a, dim=1))

print("\nTổng tích lũy theo hàng:")
print(torch.cumsum(a, dim=1))
```

## Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

### Câu a: Với `y = x²` tại `x = 3.6`

Ta có `dy/dx = 2x = 7.2` tại `x = 3.6`.

```python
import torch

x = torch.tensor(3.6, requires_grad=True)

y = x * x
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

### Câu b: Với `y = 3x² + 2x + 1`

Đạo hàm là `dy/dx = 6x + 2`. Tại `x = 3.6`, gradient bằng `23.6`.

```python
import torch

x = torch.tensor(3.6, requires_grad=True)

y = 3 * x**2 + 2 * x + 1
y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

## Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức `y = x² + 2x + 3`, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([1, 2, 3])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
print("\nMSE cuối cùng:", mse.item())
```

## Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

torch.manual_seed(42)

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1**2 + y2**2 + y3**2 + y4**2

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())

print("\nKiểm tra nghiệm:")
print("A + B =", (A + B).item())
print("C - D =", (C - D).item())
print("A + C =", (A + C).item())
print("B - D =", (B - D).item())
```

#### Yêu cầu 3: Cho `y = 2x² + 5x + 3`. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

y = 2 * x**2 + 5 * x + 3
y.backward()

print("y =", y.item())
print("dy/dx =", x.grad.item())
```

#### Yêu cầu 4: Thay đa thức trong Bài 9 bằng `y = 2x² - 3x + 5`; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

np.random.seed(42)
torch.manual_seed(42)

polynomial = np.poly1d([2, -3, 5])

N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

w = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(2000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean((y - y_pred)**2)
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
print("\nMSE cuối cùng:", mse.item())
```
